## 1 Importar librerías

```python
# Data manipulation
import numpy as np
import pandas as pd

# Interactive plotting
import plotly.express as px
import plotly.graph_objects as go

# Jupyter Notebook interactive widgets & rendering
import ipywidgets as widgets
from IPython.display import HTML, clear_output, display
```

## 2 Generar y cargar el dataset con sprites

Este notebook crea un archivo `PokemonSprites.csv` a partir del dataset original `Pokemon.csv`, añadiendo una columna `Sprite` con la URL del sprite oficial de PokéAPI.

```python
# Cargar el dataset original y generar el CSV con sprites
pokemon_raw = pd.read_csv("PokemonSprites.csv")
pokemon_raw.columns = pokemon_raw.columns.str.strip()

id_column = "ID" if "ID" in pokemon_raw.columns else "#"

pokemon_raw["Sprite"] = (
    "https://raw.githubusercontent.com/PokeAPI/sprites/master/sprites/pokemon/"
    + pokemon_raw[id_column].astype(str)
    + ".png"
)

pokemon_raw.to_csv("PokemonSprites.csv", index=False)
print("Archivo PokemonSprites.csv generado con", len(pokemon_raw), "registros.")

# Cargar el dataset con la nueva columna
pokemon = pd.read_csv("PokemonSprites.csv")
```

## 3 Descripción

- Origen: `Pokemon.csv` del repositorio `pokemonData`
- Número de registros: total de Pokémon cargados

- Generaciones: conteo por generación

```python
print("Registros:", pokemon.shape[0])
print("Columnas:", pokemon.columns.tolist())
print("Generaciones:")
print(pokemon["Generation"].value_counts().sort_index())
```

## 4 Inspección

```python
display(pokemon.head())
print("Shape:", pokemon.shape)
print("Info:")
print(pokemon.info())
print("Descripción estadística:")
print(pokemon.describe(include=[np.number]))
```

## 5 Limpiar columnas

```python
pokemon.columns = (
    pokemon.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_", regex=False)
)
print(pokemon.columns.tolist())
```

```python
pokemon["type1"] = pokemon["type1"].str.title()
pokemon["type2"] = pokemon["type2"].str.title()
print(pokemon[["type1", "type2"]].head(10))
```

## 14 Scatter Plot 3D

```python
import plotly.express as px

#=========================================
# Preparación del DataFrame
#=========================================

pokemon = pokemon.copy()

pokemon = pokemon.sort_values(
    ["generation","type1","promedio_estadisticas"]
)

pokemon["generation"] = pokemon["generation"].astype(str)

#=========================================
# Scatter Plot 3D
#=========================================

fig = px.scatter_3d(

    pokemon,

    x="generation",
    y="type1",
    z="promedio_estadisticas",

    color="type1",

    size="promedio_estadisticas",
    size_max=18,

    opacity=0.85,

    hover_name="name",

    custom_data=[
        "sprite",
        "hp",
        "attack",
        "defense",
        "sp._atk",
        "sp._def",
        "speed",
        "total"
    ],

    color_discrete_sequence=px.colors.qualitative.Dark24,

    labels={
        "generation":"Generación",
        "type1":"Tipo principal",
        "promedio_estadisticas":"Promedio de estadísticas"
    },

    template="plotly_dark",

    height=850,
    width=1400
)
# ===========================
# Filtros
# ===========================

generaciones = ["Todas"] + sorted(pokemon["generation"].astype(str).unique().tolist())

tipos = ["Todos"] + sorted(pokemon["type1"].unique().tolist())

generacion = widgets.Dropdown(
    options=generaciones,
    value="Todas",
    description="Generación:"
)

tipo = widgets.Dropdown(
    options=tipos,
    value="Todos",
    description="Tipo:"
)

promedio = widgets.IntRangeSlider(
    value=[
        int(pokemon["promedio_estadisticas"].min()),
        int(pokemon["promedio_estadisticas"].max())
    ],
    min=int(pokemon["promedio_estadisticas"].min()),
    max=int(pokemon["promedio_estadisticas"].max()),
    step=1,
    description="Promedio:",
    continuous_update=False,
    layout=widgets.Layout(width='700px')
)

nombre = widgets.Text(
    value="",
    placeholder="Ejemplo: Pikachu",
    description="Buscar:"
)

orden = widgets.Dropdown(
    options=[
        "Promedio",
        "HP",
        "Ataque",
        "Defensa",
        "Velocidad",
        "Total"
    ],
    value="Promedio",
    description="Orden:"
)


widgets.interactive_output(

    actualizar,

    {
        "gen":generacion,
        "tip":tipo,
        "rango":promedio,
        "buscar":nombre,
        "ordenar":orden
    }

)

display(

    widgets.VBox([

        generacion,

        tipo,

        promedio,

        nombre,

        orden

    ])

)
#=========================================
# Apariencia de los puntos
#=========================================

fig.update_traces(

    marker=dict(

        line=dict(
            color="white",
            width=0.4
        )
    ),

    hovertemplate="""

<b style='font-size:20px'>%{hovertext}</b>

<hr>

<b>Generación:</b> %{x}

<b>Tipo:</b> %{y}

<b>Promedio:</b> %{z:.2f}

<hr>

<b>HP:</b> %{customdata[1]}

<b>Ataque:</b> %{customdata[2]}

<b>Defensa:</b> %{customdata[3]}

<b>Ataque Especial:</b> %{customdata[4]}

<b>Defensa Especial:</b> %{customdata[5]}

<b>Velocidad:</b> %{customdata[6]}

<b>Total:</b> %{customdata[7]}

<hr>

<b>Sprite Oficial</b>

%{customdata[0]}

<extra></extra>

"""
)

#=========================================
# Diseño
#=========================================
def actualizar(gen, tip, rango, buscar, ordenar):

    df = pokemon.copy()

    if gen != "Todas":
        df = df[df["generation"].astype(str) == gen]

    if tip != "Todos":
        df = df[df["type1"] == tip]

    df = df[
        (df["promedio_estadisticas"] >= rango[0]) &
        (df["promedio_estadisticas"] <= rango[1])
    ]

    if buscar != "":
        df = df[
            df["name"].str.contains(buscar, case=False)
        ]

    columnas = {
        "Promedio":"promedio_estadisticas",
        "HP":"hp",
        "Ataque":"attack",
        "Defensa":"defense",
        "Velocidad":"speed",
        "Total":"total"
    }

    df = df.sort_values(columnas[ordenar], ascending=False)

    fig = px.scatter_3d(

        df,

        x="generation",
        y="type1",
        z="promedio_estadisticas",

        color="type1",

        size="promedio_estadisticas",

        size_max=18,

        opacity=0.85,

        hover_name="name",

        custom_data=[
            "sprite",
            "hp",
            "attack",
            "defense",
            "sp._atk",
            "sp._def",
            "speed",
            "total"
        ],

        color_discrete_sequence=px.colors.qualitative.Dark24,

        template="plotly_dark",

        height=850,

        width=1400
    )

    fig.update_traces(

        marker=dict(
            line=dict(color="white", width=0.4)
        ),

        hovertemplate="""
<b style='font-size:20px'>%{hovertext}</b><br><br>

<b>🎮 Generación:</b> %{x}<br>
<b>🔥 Tipo:</b> %{y}<br>
<b>⭐ Promedio:</b> %{z:.2f}<br><br>

<b>❤️ HP:</b> %{customdata[1]}<br>
<b>⚔️ Ataque:</b> %{customdata[2]}<br>
<b>🛡️ Defensa:</b> %{customdata[3]}<br>
<b>✨ At. Especial:</b> %{customdata[4]}<br>
<b>🛡️ Def. Especial:</b> %{customdata[5]}<br>
<b>💨 Velocidad:</b> %{customdata[6]}<br>
<b>📊 Total:</b> %{customdata[7]}<br><br>

<b>Sprite:</b><br>
%{customdata[0]}

<extra></extra>
"""
    )

    fig.update_layout(

        title="📊 Scatter Plot 3D Interactivo de Pokémon",

        scene=dict(

            xaxis_title="Generación",

            yaxis_title="Tipo",

            zaxis_title="Promedio"

        )
    )

    fig.show()


#=========================================
# Mostrar
#=========================================

fig.show()

#=========================================
# Exportar HTML
#=========================================

fig.write_html(
    "ScatterPlot3D_Pokemon.html",
    include_plotlyjs="cdn"
)

print("Visualización exportada correctamente.")
```
