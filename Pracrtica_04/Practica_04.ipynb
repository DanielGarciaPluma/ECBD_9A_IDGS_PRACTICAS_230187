# **Práctica 04 : Minería de Datos en Educación**

El estudiante pondra en practica la realizacion de un analisis exploratorio de datos , con multiples datasets ,así como implementara el uso de Mapas de Calor
(Heatmaps) con referencia de posicionamiento de GPS , con la finalidad de vizualizar las frecuencias y tendencias sobre los datos que requieren una referencia
geografica mundial en su contexto.

**Contexto**: DonorsChoose es una organización sin fines de lucro con sede en EE. UU. que permite a las personas donar directamente a proyectos escolares en escuelas públicas. Hasta enero de 2018, DonorsChoose.org había financiado un millón de proyectos. En el 77 % de las escuelas públicas de Estados Unidos, se ha solicitado al menos un proyecto a través de DonorsChoose.org. DonorsChoose ha compartido un conjunto completo de datos de todos los proyectos anteriores, los recursos relacionados y los metadatos de docentes, escuelas, donantes y donaciones. Estos datos pueden utilizarse para extraer información valiosa y crear una solución que permita a DonorsChoose.org desarrollar campañas de correo electrónico dirigidas a donantes anteriores, recomendando proyectos específicos para sus aulas. En este cuaderno, he realizado un análisis y una exploración exhaustivos del conjunto de datos compartido. Este cuaderno no solo proporciona información valiosa del conjunto de datos de DonorsChoose, sino que también sirve como tutorial para realizar análisis de datos, análisis exploratorio de datos y visualizaciones de datos.
